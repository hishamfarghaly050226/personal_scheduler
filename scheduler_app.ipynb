# 📅 Personal Scheduler

A point-and-click interface for the Personal Scheduler, running inside Jupyter.
It uses the same `personal_scheduler` code as the desktop app (models, SQLite database, services), so your data in `data/scheduler.db` is shared with it.

**How to use:** run the three cells below, then use the tabs. Nothing else to type.

First time only: `pip install ipywidgets matplotlib`

```python
# 1) Make the project importable (works when this notebook sits inside the personal_scheduler folder)
import sys
from pathlib import Path

root = Path.cwd()
if not (root / "services").exists() and (root / "personal_scheduler").exists():
    root = root / "personal_scheduler"
sys.path.insert(0, str(root))

%matplotlib inline
```

```python
# 2) Open the database and create the services
from database import DatabaseManager
from services import ReminderService, Scheduler
from notebook_ui import SchedulerApp, seed_demo_data

db = DatabaseManager()              # data/scheduler.db (created automatically)
scheduler = Scheduler(db)
reminders = ReminderService(scheduler)

# Optional: fill an EMPTY database with example tasks/events. Delete this line to start blank.
seed_demo_data(scheduler)
```

```python
# 3) Launch the app
app = SchedulerApp(scheduler, reminders)
app.show()
```

### Tips
- **➕ Add**: choose Task or Event. Picking a start time fills in an end time one hour later. Overlapping events are saved with a warning.
- **✅ Tasks / 📅 Events**: filter or search, pick an item from the dropdown, then press **Done / In progress / Edit / Delete**. Deleting always asks first.
- **🔔 Reminders**: notebooks have no background timer, so press **Check reminders** on the Today tab.
- Run the last cell when you are finished so the database is closed cleanly.
